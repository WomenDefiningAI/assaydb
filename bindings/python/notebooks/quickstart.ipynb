# Assay quickstart
Install the supplied platform wheel first (no Rust compiler needed). This notebook has no saved execution output. SQL runs in Assay; pandas/Polars only convert results.

```python
import assaydb
db = assaydb.connect()
db.execute('CREATE TABLE events (id INTEGER, label TEXT)')
db.execute("INSERT INTO events VALUES (9007199254740993, 'hello'), (2, 'world')")
result = db.sql('SELECT id, label FROM events ORDER BY id')
result.to_arrow()
```
